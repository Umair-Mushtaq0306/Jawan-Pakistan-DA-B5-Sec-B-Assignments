Make a class called Employee. The __init__() method for Employee should store its different attributes: name, number, address, salary, post, year of joining, education, and his year of experience.

Make a method called describe_employee() that prints the detail Information about employee, a method update_ year_of_experience(), and a method called increase_salary() that do increamnet in it's salary.

Make an instance called employee101 from your class. Print the four attri-
butes individually, and then call all it's methods for describing employee, ssalary increamnet and increase year of experience.

Create 10 different instances from the class, and call describe_employee() for each
instance. Then do some updations with it's methods.



```python
class employee:
    def __init__(self,name,number,address,salary,post,year_of_joining,education,year_of_experiance):
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.year_of_experiance = year_of_experiance
    def describe_employee(self):
        print(f"Name: {self.name}")
        print(f"Employee Number: {self.number}")
        print(f"Address: {self.address}")
        print(f"Salary: ${self.salary}")
        print(f"Post: {self.post}")
        print(f"Year of Joining: {self.year_of_joining}")
        print(f"Education: {self.education}")
        print(f"Year of Experience: {self.year_of_experiance}")
        print("-" * 20)
    def update_year_of_experiance(self,new_experiance):
        self.year_of_experiance = new_experiance
        print(f"Year of experience updated to: {self.year_of_experiance}")
    def increase_salary(self,increament_salary):
        self.salary += increament_salary
        print(f"Salary increased by ${increament_salary}. New salary: ${self.salary}")
        
```

```python
employee101 = employee("Umair Mushtaq", "EMP101", "karachi Pakistan", 50000, "Accountant", 2020, "Bachelors", 3)
```

```python
print(f"Name: {employee101.name}")
print(f"Employee Number: {employee101.number}")
print(f"Address: {employee101.address}")
print(f"Salary: ${employee101.salary}")
```

```python
employee101.describe_employee()
employee101.increase_salary(5000)
employee101.update_year_of_experiance(4)
employee101.describe_employee()
```

```python
def employee(name,number,address,salary,post,year_of_joining,education,year_of_experiance):
    return {
        "name": name,
        "number": number,
        "address": address,
        "salary": salary,
        "post": post,
        "year_of_joining": year_of_joining,
        "education": education,
        "year_of_experiance": year_of_experiance
    }

def describe_employee(employee):
    print(f"Name: {employee['name']}")
    print(f"Employee Number: {employee['number']}")
    print(f"Address: {employee['address']}")
    print(f"Salary: ${employee['salary']}")
    print(f"Post: {employee['post']}")
    print(f"Year of Joining: {employee['year_of_joining']}")
    print(f"Education: {employee['education']}")
    print(f"Year of Experience: {employee['year_of_experiance']}")
    print("-" * 20)

def update_year_of_experiance(employee, new_experiance):
    employee['year_of_experiance'] = new_experiance
    print(f"Year of experience updated to: {employee['year_of_experiance']}")

def increase_salary(employee, increament_salary):
    employee['salary'] += increament_salary
    print(f"Salary increased by ${increament_salary}. New salary: ${employee['salary']}")
```
