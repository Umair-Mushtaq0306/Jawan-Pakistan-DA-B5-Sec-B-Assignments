```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('cars.csv')
df
```

#### Check Duplicate Records

```python
df_duplicate = df.duplicated().sum()
df_duplicate
```

#### Drop Duplicate Records

```python
df = df.drop_duplicates()
df
```

```python
df.shape
```

#### Check missing or null values

```python
df.isnull().sum()
```

#### Drop missing or null values

```python
df = df.dropna()
df.count()
```

#### Show top 10 average MSRP (Price) per car brand (rounded figure)?

```python
top_10_MSRP = df.groupby('Make')['MSRP'].mean().round().sort_values(ascending=False).head(10)
top_10_MSRP
```

#### Show the average MSRP (price) per years (rounded figure)?

```python
avg_MSRP_PerYear = df.groupby('Year')['MSRP'].mean().round().sort_index()
avg_MSRP_PerYear
```

#### Show average Engine HP (power) rounded figure and MPG (efficiency) rounded figure per year?

Hint: Average of (city mpg + highway MPG)

```python
avg_hp_mpg_per_year = df.groupby('Year').agg({
    'Engine HP': 'mean',
    'city mpg': 'mean',
    'highway MPG': 'mean'
}).round()
avg_hp_mpg_per_year
```

#### What percentage of cars are Compact, Midsize, or Large?

```python
size_per = df['Vehicle Size'].value_counts(normalize=True)*100
```

```python
Per_cars = size_per[['Compact', 'Midsize', 'Large']]
Per_cars.round(2)
```

#### Which vehicle styles are the most sold (most common) in the dataset?

```python
most_common_styles = df['Vehicle Style'].value_counts().head(10)
most_common_styles
```

#### Which car brands have the most vehicles priced at $40,000 or more and with a popularity score of at least 1000?

```python
filtered_df = df[(df['MSRP'] >= 40000) & (df['Popularity'] >= 1000)]
filtered_df
```
